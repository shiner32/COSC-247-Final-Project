```python
import pandas as pd
import yfinance as yf
from datetime import datetime
```

```python
symbols = pd.read_csv("symbols.csv")['Symbol'].tolist()
```

```python
netAssets = []
inceptionDate = []

for symbol in symbols:
    ticker = yf.Ticker(symbol)
    d = ticker.info

    netAssets.append(d.get("totalAssets"))
    inceptionDate.append(d.get("fundInceptionDate"))



df = pd.DataFrame({
    "Symbol": symbols,
    "Net Assets": netAssets,
    "Inception Date": inceptionDate
})
```

```python
netAssetsCopy = df.copy()
netAssetsCopy 
```

```python
netAssetsCopy['In Billions'] = netAssetsCopy['Net Assets'] / 1e9
netAssetsCopy.sort_values(by='Inception Date', inplace=True, ascending=False)
netAssetsCopy['Inception Date'] = pd.to_datetime(netAssetsCopy['Inception Date'], unit='s').dt.date
netAssetsCopy.to_csv("net_assets.csv", index=True)
```
